## Úkol č.3
### For cyklus a aproximace čísla $\pi$

- $\pi$, jakožto iracionální číslo (tj. s nekonečným desetinným rozvojem), nelze zapsat do počítače v celém zvém znění, přestože je nedílnou součástí výpočtů od obsahů po vlnění. Výpočty však stačí provést dostatečně přesně a k tomu potřebujeme dostatečně přesné (dostatečná přesnost závisí na typu aplikace) vyjádření čísla $\pi$. Jedna z možných aproximací je pomocí tazvané Leibnizovy řady:

$$
    \frac{\pi}{4} = \displaystyle \sum_{i=0}^n \frac{(-1)^i}{2i +1} = 1 - \frac{1}{3} + \frac{1}{5} - \frac{1}{7} + \frac{1}{9} - \frac{1}{11}\dots, \\
    n \in \mathbb{N}
$$

- Pomocí for cyklu a Leibnizovy řady aproximujte číslo $\pi$ pro $n = 5, 10, 50, 1000$.
